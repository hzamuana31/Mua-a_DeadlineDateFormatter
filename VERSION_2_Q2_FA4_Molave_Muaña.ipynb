```python
from datetime import datetime

print("=== AUTOMATED BATCH REGISTRATION FORMATTER ===")

#Places the user in a loop unless they press Q to end said loop.
while True:
    print("\n--- NEW REGISTRATION ---")
    Mfirst_name = input("Enter first name (or type 'Q' to quit): ")
    if Mfirst_name.strip().upper() == "Q":
      print("Program runtime shut down.")
      break

#User inputs for the display/output.
    Mlast_name = input("Enter last name:")
    Msection = (input("Enter section:"))
    Mevent = (input("Enter event name:"))
    yr = int(input("Event Year (YYYY): "))
    mo = int(input("Event Month (1-12): "))
    dy = int(input("Event Day (1-31): "))

    event_date = datetime(yr, mo, dy)

#Improvements: Added a date formatting and string usage.

#These clean, change and modify the inital input.
    Cfirst_name = Mfirst_name.strip().title()
    Clast_name = Mlast_name.strip().title()
    Csection = Msection.strip().upper()
    Cevent = Mevent.strip().title()
    C_yr = event_date.strftime("%Y")
    C_mo = event_date.strftime("%B")
    C_dy = event_date.strftime("%d")
    C_weekday = event_date.strftime("%A")


#Display / output of the entire program.
    print("\n=== REGISTRATION RECEIPT ===")
    print("First name:", Cfirst_name)
    print("Last name :", Clast_name)
    print("Section:", Csection)
    print("Event:", Cevent)
    print("Event date:", C_mo, C_dy, C_yr)

    print("\n EXTRA INFO:")
    print("Year:", C_yr)
    print("Month:", C_mo)
    print("Day:", C_dy)
    print("Weekday:", C_weekday)


```
